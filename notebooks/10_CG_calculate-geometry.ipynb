```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
import spatial_lite as sl

pd.set_option('display.max_columns', 40)
```

## 1. Load the data and read it as geography
Start with a quick table preview, then immediately map the administrative boundary context.

```python
df = sl.load_dataset('meuse.csv')
value = 'zinc'
lon, lat = 'lon', 'lat'
sl.quick_summary(df)
sl.map_boundaries('Calculate Geometry: administrative boundary context')
```

## 2. Map the subject layer
This map shows the teaching dataset as a spatial layer. Ask: where are high values, low values, clusters, or gaps?

```python
plot_df = df.rename(columns={lon:'lon', lat:'lat'}).copy()
sl.map_points(plot_df, color=value if value in plot_df.columns else None, title='Calculate Geometry: mapped observations')
```

## 3. Add a second layer and compare
Spatial analysis becomes clearer when layers are compared: boundaries, facilities, roads, grids, or events.

```python
df['class'] = sl.classify_quantiles(df[value], 5) if value in df.columns and pd.api.types.is_numeric_dtype(df[value]) else 'feature'
sl.map_points(df.rename(columns={lon:'lon', lat:'lat'}), color=value if value in df.columns and pd.api.types.is_numeric_dtype(df[value]) else None, title='Thematic point map')
sl.map_choropleth('rate_per_10k', 'Administrative-area context map')
if value in df.columns and pd.api.types.is_numeric_dtype(df[value]):
    display(df.groupby('class')[value].agg(['count','mean','min','max']).round(2))
```

## 4. Interactive map for exploration
The map below uses Leaflet and OpenStreetMap tiles when internet access is available. If tiles do not load, the static maps above still provide the analysis.

```python
try:
    sl.interactive_leaflet(plot_df, lon='lon', lat='lat', label=plot_df.columns[0], title='Interactive exploration map')
except Exception as exc:
    print('Interactive map skipped:', exc)
```
